```python
# Operators in python :
# Arithmetic operators :
a=10
b=20
print(a+b) # additions
print(a-b) # subtraction
print(a*b) # multiplication 
print(a/b) # division 
print(a//b) # floor division (quotient)
print(a%b) # modulo (remainder)
print(a**b) # exponential (to the power of )  
```

```python
#comparison operator:
x=20
y=30
print(x>y) # greater than
print(x>=y) # greater than and equal too 
print(x<y) # lesser than 
print(x<=y) # lesser than and equal too
print(x==y) # x is equal to y
print(x!=y) # x is not equal to y
```

```python
#logical operators: and,or,not
a=100
b=150
c=50
print(a>b and a>c) # returns true when both conditions are true; otherwise, returns false
print(a>b or a>c) # returns true when at least one condition is true 
print(not(a>b)) # returns oppostion of actual output
```

```python
# Identity operator: is,is not
a=10
print(id(a))
b=10
print(id(b))

c="python"
print(id(c))

d="python"
print(id(d))

print(c is d)
print(c is not d)
```

```python
# Membership operator : in, not in 

a=[1,2,3,4,5]
print(4 in a) # returns true if 4 is present in the list 'a', else false
print(10 not in a)
```
